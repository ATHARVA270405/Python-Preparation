```python
class Employee:
    def __init__(self,emp_id,name,salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary
    def display_details(self):
        print("Employee id : " , self.emp_id)
        print("Name is : " , self.name)
        print("Salary is : " , self.salary)
    
```

```python
class Manager(Employee):
    def __init__(self,emp_id,name,salary,dept):
        super().__init__(emp_id,name,salary)
        self.dept = dept
    def display_details(self):
        super().display_details()
        print("Department is : " , self.dept)
        

```

```python
manager = Manager(1001,"ATHARVA",25000,"MCA")
manager.display_details()
```

```python
class BankAccount:
    def __init__(self,acc_no,hold_name,balance):
        self.acc_no = acc_no
        self.hold_name = hold_name
        self.balance = balance
    def deposit(self,amt):
        self.balance+=amt
        return self.balance
    def withdraw(self,amt):
        self.balance-=amt
        return self.balance
    def display_balance(self):
        print("Balance is",self.balance)


class Saving_acc(BankAccount):
    def __init__(self,acc_no,hold_name,balance):
        super().__init__(acc_no,hold_name,balance)
        
    def withdraw(self,amt):
        if amt>self.balance:
            print("Insufficient Balance")
        return self.balance
    
        
    
        
```

```python
obj = Saving_acc(100,"ATHARVA",25000)
obj.display_balance()
```

```python
obj.withdraw(30000)
```
