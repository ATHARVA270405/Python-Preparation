# Prefix Sum

nums = [1,7,3,6,5,6]
def calculate_pivot(nums):
    total = sum(nums)
    left_sum = 0
    for i , num in enumerate(nums):
        if left_sum ==(total-left_sum-num):
            return i
        left_sum+=num
    return -1

```python
calculate_pivot([1,7,3,6,5,6])
```

# Kadane's Algorithm

```python
def max_sum_array(nums):
    current_max = nums[0]
    global_max = nums[0]
    for i in range(1,len(nums)):
            current_max = max(nums[i] , current_max+nums[i])
    if current_max > global_max:
                global_max = current_max
    return global_max
    
```

```python
max_sum_array([-2,1,-3,4,-1,2,1,-5,4])
```
