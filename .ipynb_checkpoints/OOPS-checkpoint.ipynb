```python
class BankAccount:
    def __init__(self,acc_no,name,balance):
        self.acc_no = acc_no
        self.__name = name
        self.__balance = balance

    def deposit(self,amt):
        if amt <0:
            print("Enter the Positive Amount")
        else:
            self.__balance+=amt
            return self.__balance

    def withdraw(self,amt):
        if amt>self.__balance:
            print("Amount is greater than Balance")
        else:
            self.__balance-=amt
            return self.__balance
    def display_balance(self):
        print("The Balance is",self.__balance)
    
```

```python
customer1 = BankAccount(1001,"Akash",50000)
```

```python
customer1.display_balance()
```

```python
customer1.deposit(5000)
```

```python
customer1.withdraw(10000)
```

```python
customer1.display_balance()
```

```python
customer1.deposit(-24924)
```

```python
customer1.withdraw(50000)
```

```python
class student :
    def __init__(self,roll_no,name,*marks):
        self.roll_no = roll_no
        self.name = name
        self.marks= marks

    def total_marks(self):
        sum1 = 0
        for mark in self.marks:
            sum1+=mark
        return sum1

    def percentage(self):
        total = self.total_marks()
        percent = total/len(self.marks)
        return percent

    def grade(self):
        if self.percentage()>= 90:
            return "+A"
        elif self.percentage()>= 80 and self.percentage()< 90:
            return "B+"
        elif self.percentage()>= 70 and self.percentage()< 80:
            return "C+"
        elif self.percentage()>= 60 and self.percentage()< 70:
            return "D+"
        else :
            return "Fail"

    def display_report(self):
        print("Name is :",self.name)
        print("Roll no is :",self.roll_no)
        print("Total Marks obtained is :",self.total_marks())
        print("Percentage obtained is :",self.percentage())
        print("Grade is :",self.grade())
        
        
    
        
        
        
```

```python
student1 = student(10,"Atharva",89,34,67,89,43)
```

```python
student1.total_marks()
```

```python
student1.percentage()
```

```python
student1.grade()
```

```python
student1.display_report()
```
