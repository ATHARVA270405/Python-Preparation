# Sliding Window 

# Maximum Average Subarray I (Easy) LeetCode : 643

```python
def max_avg_subarray(arr,k):
    curr_sum = sum(arr[:k])
    maxx = curr_sum
    for i in range(k,len(arr)):
        curr_sum += arr[i] - arr[i-k]
        if curr_sum>maxx:
            maxx = curr_sum
    return float(maxx)/k
```

```python
max_avg_subarray([1,12,-5,-6,50,3],4)
```
