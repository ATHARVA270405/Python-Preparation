```python
name = input("Enter the name=>")
physics_marks = int(input("Enter the physics_marks =>"))
chemitry_marks = int(input("Enter the chemitry_marks =>"))
maths_marks = int(input("Enter the maths_marks =>"))
attendance = int(input("Enter the attendance"))
total_marks = physics_marks+chemitry_marks+maths_marks
avg = float(total_marks)/3
percentage = (float(total_marks)/300)*100
if percentage<60:
    grade = "Fail"
elif percentage>=60 and percentage<70:
    grade = "C"
elif percentage>=70 and percentage<80:
    grade = "B"
elif percentage>=80 and percentage<90:
    grade = "A"
else:
    grade = "+A"
if attendance>=75:
    isEligible = "Eligible"
else:
    isEligible = "Not Eligible"
print("Name",name)
print("Total Marks",total_marks)
print("Average",avg)
print("Grade",grade)
print("Eligible for Exam?",isEligible)
```
