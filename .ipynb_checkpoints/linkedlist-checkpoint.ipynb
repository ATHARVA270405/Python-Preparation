# LinkedList

```python
#creating the node of the LinkedList
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
```

```python
#create the object of the linked list
#also the nodes are connected to each other
head = Node(10)
head.next = Node(20)
head.next.next = Node(30)
head.next.next.next = Node(40)
```

```python
print(head.data)
```

```python
print(head.next.data)
```

```python
print(head.next.next.data)
```

```python
#Instead of doing these... we can first create the objects then connect them manually
#creating the node of the LinkedList
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
```

```python
#here we created the ojects
first = Node(10)
second = Node(20)
third = Node(30)
fourth = Node(40)
```

```python
print(first.next)
#because we arent connect the nodes yet
```

```python
#here we connect the nodes
first.next = second
second.next = third
third.next = fourth
```

```python
print(first.next)
#it gives the address of the second node
```

# Traversal

```python
current = first
while current :
    print(current.data)
    current = current.next
```

# Insert The Node in the Middle

```python
# create the inserted Node object
mid = Node(25)
```

```python
#we want to add these in the middle 
mid.next = third
second.next = mid
```

```python
current = first
while current :
    print(current.data)
    current = current.next
#see here , the node added in the middle , but these is specific .... position we took
```

```python
def insert(first, value,new_value):
    current = first
    while current:
        if current.data == value:
            #create the node
            new_node = Node(new_value)
            new_node.next = current.next
            current.next = new_node
            return first
        current = current.next
    return first
            
        
            
```

```python
first = insert(first,30,35)
```

```python
current = first
while current :
    print(current.data)
    current = current.next
#here we added the 35 as on the required position...
```
