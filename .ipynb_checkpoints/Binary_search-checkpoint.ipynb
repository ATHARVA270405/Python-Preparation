# Binary Search

```python
def binary_search(arr,target):
    left = 0
    right = len(arr)-1
    while left < right:
        mid = (left+right)//2
        if target == arr[mid]:
            return mid
        elif target >arr[mid]:
            left = mid+1
        else :
            right = mid-1
    return -1
```

```python
binary_search([-1,0,3,5,9,12],9)
```
