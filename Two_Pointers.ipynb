# Two Pointers

```python
import re
def palimdrome(str):
    str1 = re.sub(r'[^a-zA-Z0-9]','',str).lower()
    left = 0
    right = len(str1)-1
    while left<right:
        if str1[left] != str1[right]:
            return False
        left+=1
        right-=1
    return True
palimdrome("I am Brouhhhhhh")
palimdrome(" I am ma I")
```

```python
def two_sum(arr,target):
    left = 0
    right = len(arr)-1
    while left < right :
        curr_sum = arr[left]+ arr[right]
        if curr_sum == target:
            return [left+1,right+1]
        elif curr_sum<target:
            left+=1
        else:
            right-=1
```

```python
two_sum([2,7,11,15],18)
```
