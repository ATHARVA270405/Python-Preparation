```python
class Payment:
    def pay(self,amt):
        pass

class CreditCard(Payment):
    def pay(self,amt):
        print("Payment success by using credit card")

class UPI(Payment):
    def pay(self,amt):
        print("Payment success by using UPI")

class NetBanking(Payment):
    def pay(self,amt):
        print("Payment success by using NetBanking")

payment = [CreditCard(),UPI(),NetBanking()]
for i in payment:
    i.pay(1000)
        
```
